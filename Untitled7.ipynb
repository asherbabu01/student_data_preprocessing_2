```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("University of Regina.csv")

print(df.head())
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print(df.columns)
```

```python
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python
df.columns = df.columns.str.strip()

print(df.columns)
```

```python
numeric_columns = [
    "Current Salary",
    "Administrative/ Research Stipend",
    "Market Supplement",
    "Total"
]
```

```python
for column in numeric_columns:
    df[column] = (
        df[column]
        .astype(str)
        .str.replace(",", "", regex=False)
        .replace("nan", np.nan)
    )

    df[column] = pd.to_numeric(df[column], errors="coerce")
```

```python
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
df["Administrative/ Research Stipend"] = \
    df["Administrative/ Research Stipend"].fillna(0)

df["Market Supplement"] = \
    df["Market Supplement"].fillna(0)
```

```python
print(df.isnull().sum())
```

```python
df["Last Name"] = df["Last Name"].str.strip()
df["First Name"] = df["First Name"].str.strip()
```

```python
print(df[["First Name", "Last Name"]].head())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print(df[df["Current Salary"] < 0])
```

```python
for column in numeric_columns:
    print(column)
    print(df[df[column] < 0])
```

```python
for column in numeric_columns:
    df.loc[df[column] < 0, column] = np.nan
```

```python
print(df[numeric_columns].describe())
```

```python
print("Average salary:",
      df["Current Salary"].mean())

print("Maximum salary:",
      df["Current Salary"].max())

print("Minimum salary:",
      df["Current Salary"].min())
```

```python
Q1 = df["Current Salary"].quantile(0.25)
Q3 = df["Current Salary"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
outliers = df[
    (df["Current Salary"] < lower_limit) |
    (df["Current Salary"] > upper_limit)
]

print(outliers)
```

```python
print("Number of outliers:", len(outliers))
```

```python
salary = df["Current Salary"]

df["Salary_Normalized"] = (
    salary - np.min(salary)
) / (
    np.max(salary) - np.min(salary)
)
```

```python
print(df[["Current Salary", "Salary_Normalized"]].head())
```

```python
df["Salary_Standardized"] = (
    df["Current Salary"] - np.mean(df["Current Salary"])
) / np.std(df["Current Salary"])
```

```python
print(df[["Current Salary", "Salary_Standardized"]].head())
```

```python
df["Additional_Pay"] = (
    df["Administrative/ Research Stipend"] +
    df["Market Supplement"]
)
```

```python
df["Calculated_Total"] = (
    df["Current Salary"] +
    df["Additional_Pay"]
)
```

```python
print(
    df[
        [
            "Current Salary",
            "Additional_Pay",
            "Total",
            "Calculated_Total"
        ]
    ].head()
)
```

```python
df["Difference"] = (
    df["Total"] - df["Calculated_Total"]
)

print(df["Difference"].describe())
```

```python
df_sorted = df.sort_values(
    by="Current Salary",
    ascending=False
)

print(
    df_sorted[
        ["First Name", "Last Name", "Current Salary"]
    ].head(10)
)
```

```python
df.to_csv(
    "University_of_Regina_Preprocessed.csv",
    index=False
)
```
